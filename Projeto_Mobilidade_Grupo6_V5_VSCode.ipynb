## 3.4. Heatmap de Conflitos Potenciais (CORRIGIDO PARA VSCODE)
Este gráfico cruza os horários com a **queda de velocidade** (Velocidade Regul - Velocidade Real).

```
# Cálculo da queda de velocidade (Congestionamento)
df['queda_velocidade'] = df['Velocidade Regul'] - df['Velocidade 1']
df['queda_velocidade'] = df['queda_velocidade'].clip(lower=0) # Apenas quedas positivas

# Agrupando para garantir que o heatmap tenha dados consistentes
heatmap_data = df.groupby('hora')['queda_velocidade'].mean().reset_index()

# Criando o Heatmap usando Plotly Density Heatmap mas com lógica simplificada para VSCode
fig_heat = px.density_heatmap(df, x='hora', y='Velocidade 1', 
                            z='queda_velocidade', 
                            histfunc='avg',
                            title='Mapa de Calor: Intensidade de Conflitos (Hora vs Velocidade)',
                            labels={'hora': 'Hora do Dia', 'Velocidade 1': 'Velocidade Real', 'queda_velocidade': 'Nível de Atraso'},
                            color_continuous_scale='YlOrRd')

# Ajuste para renderização no VSCode/Jupyter
fig_heat.update_layout(margin=dict(l=20, r=20, t=40, b=20))
fig_heat.show()
```

### Por que este gráfico é disruptivo?
Ele não mostra apenas onde o trânsito está lento, mas **quão longe da velocidade ideal** a via está operando em cada faixa horária, permitindo identificar o 'custo de oportunidade' da mobilidade urbana.
